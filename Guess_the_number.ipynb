```python
import random
guess=random.randint(0,10)
n=1
while (n<=3):
    user=int(input('Guess the number'))
    n+=1
    if (user==guess):
          print(f'You Guessed Right number is {guess}')
    elif(user<guess):
          print('Try Guessing Higher Number')
    else:
        print('Try Guessing Lower Number')
    break
```
