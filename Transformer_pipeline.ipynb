```python
from transformers import pipeline
```

```python
classifier=pipeline('sentiment-analysis')
```

```python
result=classifier('I Love AI')
print(result)
```

```python
def sa(sent):
  classifier=pipeline('sentiment-analysis')
  result=classifier(sent)
  return result
```

```python
pip install gradio
```

```python
import gradio as gr

def analyze_sentiment(text):
    res = sa(text)
    label = res[0]['label']
    score = res[0]['score'] * 100
    return f"{label} (Confidence: {score:.2f}%)"

demo = gr.Interface(
    fn=analyze_sentiment,
    inputs=gr.Textbox(lines=2, placeholder="Type your text here to analyze sentiment..."),
    outputs="text",
    title="Sentiment Analysis App",
    description="Enter a sentence and find out if it's positive or negative using Hugging Face pipelines."
)

demo.launch(share=True)
```

```python
gen=pipeline('text-generation',model='distilgpt2')
```

```python
result=gen('Machine Learning is',max_length=10)
```

```python
def generate_text(prompt):
    # Generate text using the 'gen' pipeline
    response = gen(prompt, max_new_tokens=50)
    return response[0]['generated_text']

# Set up the Gradio interface for text generation
text_gen_demo = gr.Interface(
    fn=generate_text,
    inputs=gr.Textbox(lines=2, placeholder="Enter a prompt to start generating text..."),
    outputs="text",
    title="Text Generation App",
    description="Enter a prompt and watch DistilGPT2 autocomplete your text!"
)

# Launch the app
text_gen_demo.launch(share=True)
```
